# A robot navigation example through observation-posterior 2BSDE method.

The robot is a point mass moving in a 2D plane $E:=[-4,4]^2$. It starts at $y_0$. It wants to search a target $y_{\rm target}$. 

The plane $E$ can be eqaully divided into $64$ square cells. There will be no obstacles in the outer $28$ cells, and $y_0$ is located in the outer cells and $y_{\rm target}$ is located in the inner cells. 

The random variable $X$ denotes which inner cell contains $y_{\rm target}$, which is a $6\times 6$ random binary matrix. 

We receive a time-continuous noisy observation $Y^1=(Y^1_t)_{t \in [0,T]}$ of the robot's location, for example, a GPS signal. The $Y^1$ is $\mathbb{R}^2$-valued. 

We also receive a time-continuous noisy observation $Y^2=(Y^2_t)_{t \in [0,T]}$ of the robot's sensor, for example, a radar signal. 

The robot will be controlled by $u$, which will control the robot's velocity.
